# Activity 3: Colab–VS Code Bridge

**Question:** What percentage of these five synthetic transactions is flagged for review?

**Data provenance:** The five records below are invented for this exercise. They contain no personal or real financial information.

**Expected output:** A five-row transaction table and the percentage flagged for review.

Run every cell from a fresh Google Colab runtime. This notebook uses only Python and pandas and does not read local files.

```python
import sys
import pandas as pd

print(f"Python: {sys.version.split()[0]}")
print(f"pandas: {pd.__version__}")
```

```python
transactions = pd.DataFrame({
    "transaction_id": ["T-001", "T-002", "T-003", "T-004", "T-005"],
    "amount": [120.00, 950.00, 45.50, 2300.00, 78.25],
    "flagged_for_review": [False, True, False, True, False],
})
transactions
```

```python
total_transactions = len(transactions)
flagged_count = int(transactions["flagged_for_review"].sum())
flagged_percentage = flagged_count / total_transactions * 100

print(f"Transactions: {total_transactions}")
print(f"Flagged for review: {flagged_count}")
print(f"Flagged percentage: {flagged_percentage:.1f}%")
```

## Conclusion

In this invented sample, 2 of 5 transactions are flagged for review (40%). This is a descriptive calculation only: the flags were supplied in the synthetic data, so the notebook does not detect fraud or establish a real-world fraud rate.
